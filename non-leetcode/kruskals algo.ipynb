```python
class dsu:
    def __init__(self,n,nodes):
        self.g=[]
        self.n=n
        self.e=0
        self.r={i:0 for i in nodes} 
        self.nodes=nodes
        self.p={i:i for i in nodes}
        self.c=0

    def _find(self,x):
        if self.p[x]!=x:
            self.p[x]=self._find(self.p[x])
        return self.p[x]

    def _union(self,x,y,c):
        px=self._find(x)
        py=self._find(y)

        if py==px:
            print("nope")
            return
        self.c+=c
        if self.r[px]>=self.r[py]:
            self.p[py]=px
            self.r[px]+=1
        else:
            self.p[px]=py
            self.r[py]+=1
        self.e+=1
        self.c+=int(c)
        self.g.append([x,y,c])
    
    def add(self):
        gg=int(input("enter no of edges:"))
        l=[]
        for _ in range(gg):
            x,y,c=input().split()
            l.append([x,y,int(c)])
        l.sort(key=lambda x:(x[-1]))
        for x,y,c in l:
            self._union(x,y,c)
        print(self.e)
        print(self.g)
    


```

```python
d=dsu(5,['a',"b",'c','d','e'])
```
