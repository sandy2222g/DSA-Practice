```python
from collections import defaultdict
import heapq

class Graph:
    def __init__(self, n):
        self.n = n
        self.g = defaultdict(list)

    def add(self, edges):
        for u, v, w in edges:
            self.g[u].append((v, w))
            self.g[v].append((u, w))

    def prim(self, start):
        visited = [False] * self.n
        pq = [(0, start)]
        mst = 0

        while pq:
            wt, node = heapq.heappop(pq)

            if visited[node]:
                continue

            visited[node] = True
            mst += wt

            for nei, w in self.g[node]:
                if not visited[nei]:
                    heapq.heappush(pq, (w, nei))

        return mst
```
